```python
#We will use  only the large urls
books = books[['ISBN', 'Book-Title', 'Book-Author', 'Year-Of-Publication', 'Publisher', 'Image-URL-L']]
books.head(10)
```

```python
books.rename(
    columns = {
        "Book-Title": "title",
        "Book-Author": "author",
        "Year-Of-Publication": "publish_year",
        "Publisher": "publisher",
        "Image-URL-L": "image_url"
    },
    inplace = True
)
books.head()
```

```python
users = pd.read_csv('BX-Users.csv', sep=";", on_bad_lines="skip", encoding='latin-1')  #ignore errors in some lines and skip them. Options - skip, warn, error.
```

```python
users.head()
```

```python
users.shape
```

```python
users.rename(
    columns = {
        "User-ID": "user_id",
        "Location": "location",
        "Age": "age"
    },
    inplace = True
)
users.head()
```

```python
book_ratings = pd.read_csv('BX-Book-Ratings.csv', sep=";", on_bad_lines="skip", encoding='latin-1')  #ignore errors in some lines and skip them. Options - skip, warn, error.
book_ratings.head()
```

```python
book_ratings.rename(
    columns = {
        "User-ID": "user_id",
        "Book-Rating": "book_rating"
    },
    inplace = True
)
book_ratings.head()
```

```python
print(books.shape, users.shape, book_ratings.shape, sep='\n')
```

```python
book_ratings['user_id'].value_counts()
```

```python
# we will filter out users who have read more than 200 books
# this is a part of feature engineering
x = book_ratings['user_id'].value_counts() > 200
x
```

```python
x[x].shape
```

```python
y = x[x].index
y
```

```python
book_ratings = book_ratings[book_ratings['user_id'].isin(y)]
book_ratings.head()
```

```python
# now merge book_ratings with books 
books_with_ratings = books.merge(book_ratings, on="ISBN")
books_with_ratings.head()
```

```python
# we want to find out which book has got how many ratings
# we apply group by operation
rating_count = books_with_ratings.groupby('title')['book_rating'].count().reset_index()
```

```python
rating_count.head()
```

```python
rating_count.rename(columns={
    "book_rating": "rating_count"
}, inplace=True)
rating_count.head()
```

```python
books_rate_count = books_with_ratings.merge(rating_count, on="title")
books_rate_count.head()
```

```python
# we will only take the books with >=50 ratings
books_rate_count = books_rate_count[books_rate_count['rating_count']>=50]
books_rate_count.head()
```

```python
books_rate_count.shape
```

```python
# dropping duplicates
books_rate_count.drop_duplicates(['user_id', 'title'], inplace=True)
books_rate_count.shape
```

```python
# check for missing values
books_rate_count.isnull().sum()
```

```python
# for collaborative filtering, convert dataframe into  pivot table
book_pivot = books_rate_count.pivot_table(columns='user_id', index='title', values='book_rating')
book_pivot.head()
```

```python
# convert NAN values to 0
book_pivot.shape
```

```python
book_pivot.fillna(0,inplace=True)
book_pivot.head()
```

```python
# before training the model, convert pivot table into CSR metrics
from scipy.sparse import csr_matrix
book_sparse = csr_matrix(book_pivot)
book_sparse
```
